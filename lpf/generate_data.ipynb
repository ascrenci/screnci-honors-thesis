```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def genRC(fc):
    R = np.random.uniform(1000,50000)
    C = 1 / (2*np.pi*R*fc)

    return R, C
```

Output will be 3D array, where each slice contains X and Y

[[R, mag...],
 [C, phase ...]]

array will be shape:
(100 fc's * 9 decades * 10 RC per fc, 2 features, ac sweep length + 1 for R and C to be stored in first column)

```python
# fc array, log spaced points from 10 Hz to 1 GHz, 100 points per decade
fc_arr = np.logspace(np.log10(10), np.log10(5E9), 9*400)
f = np.logspace(np.log10(10), np.log10(10E9), 9*200)

# initializes empty array, shape explained above
data = np.empty((400*9*1, 2, 9*200+1))
count = 0

for i,fc in enumerate(fc_arr):
    print(f"Generating for fc = {fc} ({i+1}/{len(fc_arr)})")

    for j in range(1):
        # Generate R,C, and ac sweep
        R,C = genRC(fc)
        hf = 1 / (1 + 1j*2*np.pi*f*R*C)

        # Get mag, phase plots
        #mag = 20*np.log10(np.abs(hf))
        #mag /= np.abs(mag).max() # normalize dB plot
        #phase = np.angle(hf)/np.pi # normalize phase to [-1, 0]

        # Removed normalization and dB scaling, would rather model do that work
        mag = np.abs(hf)
        phase = np.angle(hf)

        # store in array
        data[count,:,0] = np.array([fc, R*C])
        data[count,0,1:] = mag
        data[count,1,1:] = phase
        count += 1

np.save("data.npy", data)
```
